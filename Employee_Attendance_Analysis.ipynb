## 1. Import the libraries

```python
import pandas as pd
import random

random.seed(10)
```

## 2. Create the sample employee dataset

Each row represents one employee for one attendance period. Attendance values are sample monthly totals. Attrition means the employee has left the company (`Yes`) or is still working (`No`).

```python
names = [
    "Aarav", "Vivaan", "Aditya", "Vihaan", "Arjun", "Sai", "Reyansh", "Ayaan",
    "Krishna", "Ishaan", "Ananya", "Diya", "Aadhya", "Saanvi", "Pari", "Myra",
    "Ira", "Anika", "Riya", "Meera", "Kabir", "Rohan", "Rahul", "Priya",
    "Amit", "Neha", "Karan", "Riya", "Vikas", "Sneha"
]

departments = ["IT", "HR", "Finance", "Sales", "Marketing"]
roles = ["Executive", "Associate", "Senior Associate", "Manager"]
data = []

for i in range(1, 81):
    name = random.choice(names) + str(i)
    department = random.choice(departments)
    role = random.choice(roles)
    experience = random.randint(1, 15)
    working_days = 22
    present_days = random.randint(15, 22)
    absent_days = working_days - present_days
    leave_days = random.randint(0, 3)
    satisfaction = random.randint(1, 5)
    overtime = random.choice(["Yes", "No"])
    attrition = random.choices(["Yes", "No"], weights=[25, 75])[0]

    data.append({
        "Employee_ID": i,
        "Name": name,
        "Department": department,
        "Role": role,
        "Experience": experience,
        "Working_Days": working_days,
        "Present_Days": present_days,
        "Absent_Days": absent_days,
        "Leave_Days": leave_days,
        "Satisfaction": satisfaction,
        "Overtime": overtime,
        "Attrition": attrition
    })

df = pd.DataFrame(data)
df.head()
```

## 3. Understand the dataset

```python
print("First 5 records:")
display(df.head())

print("Last 5 records:")
display(df.tail())

print("Number of rows and columns:", df.shape)

print("Column names:")
print(df.columns)

print("Data types:")
print(df.dtypes)

print("Basic numerical summary:")
display(df.describe())
```

## 4. Check and clean the data

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("Number of duplicate rows:")
print(df.duplicated().sum())

# Create a duplicate row only to demonstrate duplicate removal
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print("Duplicate rows after adding one practice duplicate:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicate rows:")
print(df.shape)

# Add a few missing satisfaction values to demonstrate missing-value handling
df.loc[[5, 17, 31], "Satisfaction"] = pd.NA

print("Missing values after adding practice missing values:")
print(df.isnull().sum())

# Fill missing satisfaction values with the column mean
df["Satisfaction"] = df["Satisfaction"].fillna(df["Satisfaction"].mean())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 5. Add attendance percentage

```python
df["Attendance_Percentage"] = (df["Present_Days"] / df["Working_Days"]) * 100

print("Employee attendance details:")
display(df[["Employee_ID", "Department", "Present_Days", "Absent_Days", "Attendance_Percentage"]].head())

print("Average attendance percentage:", round(df["Attendance_Percentage"].mean(), 2), "%")
print("Average absent days:", round(df["Absent_Days"].mean(), 2))
```

## 6. Employee count and department distribution

```python
total_employees = len(df)
print("Total employees:", total_employees)

print("Employees in each department:")
print(df["Department"].value_counts())

print("Number of different departments:", df["Department"].nunique())
```

## 7. Overall attrition analysis

```python
attrition_count = df["Attrition"].value_counts()

employees_left = len(df[df["Attrition"] == "Yes"])
employees_stayed = len(df[df["Attrition"] == "No"])

attrition_rate = (employees_left / total_employees) * 100

print("Employees who left:", employees_left)
print("Employees who stayed:", employees_stayed)
print("Overall attrition rate:", round(attrition_rate, 2), "%")
print("Attrition category counts:")
print(attrition_count)
```

## 8. Department-wise attrition analysis

```python
print("Department-wise employee and attrition details")

for department in df["Department"].unique():
    department_data = df[df["Department"] == department]

    department_total = len(department_data)
    department_left = len(department_data[department_data["Attrition"] == "Yes"])
    department_rate = (department_left / department_total) * 100

    print("--------------------------------")
    print("Department:", department)
    print("Total employees:", department_total)
    print("Employees who left:", department_left)
    print("Attrition rate:", round(department_rate, 2), "%")
```

## 9. Experience group and attrition

```python
# Create an experience group using if-else
experience_group = []

for experience in df["Experience"]:
    if experience <= 3:
        experience_group.append("0-3 years")
    elif experience <= 7:
        experience_group.append("4-7 years")
    else:
        experience_group.append("Above 7 years")

df["Experience_Group"] = experience_group

print("Employees in each experience group:")
print(df["Experience_Group"].value_counts())

print("Attrition rate by experience group:")
for group in df["Experience_Group"].unique():
    group_data = df[df["Experience_Group"] == group]
    group_total = len(group_data)
    group_left = len(group_data[group_data["Attrition"] == "Yes"])
    group_rate = (group_left / group_total) * 100

    print(group, "->", round(group_rate, 2), "%")
```

## 10. Satisfaction comparison: employees who left and stayed

```python
left_data = df[df["Attrition"] == "Yes"]
stayed_data = df[df["Attrition"] == "No"]

print("Average satisfaction of employees who left:",
      round(left_data["Satisfaction"].mean(), 2))

print("Average satisfaction of employees who stayed:",
      round(stayed_data["Satisfaction"].mean(), 2))

print("Satisfaction counts for employees who left:")
print(left_data["Satisfaction"].value_counts().sort_index())

print("Satisfaction counts for employees who stayed:")
print(stayed_data["Satisfaction"].value_counts().sort_index())
```

## 11. Overtime and attendance comparison

```python
overtime_data = df[df["Overtime"] == "Yes"]
no_overtime_data = df[df["Overtime"] == "No"]

print("Employees doing overtime:", len(overtime_data))
print("Average attendance for overtime employees:",
      round(overtime_data["Attendance_Percentage"].mean(), 2), "%")

print("Employees not doing overtime:", len(no_overtime_data))
print("Average attendance for employees not doing overtime:",
      round(no_overtime_data["Attendance_Percentage"].mean(), 2), "%")

print("Attrition among employees doing overtime:")
print(overtime_data["Attrition"].value_counts())

print("Attrition among employees not doing overtime:")
print(no_overtime_data["Attrition"].value_counts())
```

## 12. Find employees with highest and lowest attendance

```python
print("Top 5 attendance percentages:")
display(df.nlargest(5, "Attendance_Percentage")[
    ["Employee_ID", "Department", "Attendance_Percentage"]
])

print("Bottom 5 attendance percentages:")
display(df.nsmallest(5, "Attendance_Percentage")[
    ["Employee_ID", "Department", "Attendance_Percentage"]
])
```

## 13. Create a summary using a user-defined function

```python
def show_department_summary(department_name):
    department_data = df[df["Department"] == department_name]

    total = len(department_data)
    left = len(department_data[department_data["Attrition"] == "Yes"])
    average_attendance = department_data["Attendance_Percentage"].mean()

    print("Department:", department_name)
    print("Total employees:", total)
    print("Employees who left:", left)
    print("Average attendance:", round(average_attendance, 2), "%")

show_department_summary("IT")
```
